```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import *
from sympy import symbols, approximants
from sympy.physics.quantum import TensorProduct
import sympy as sp
from sympy import symbols

# Define symbols
E0, E1, E2 = 0, 0, 0
J, w = symbols('J epsilon', real=True, positive=True)

# Define Matrix
H = sp.Matrix([
    [0, J, J, J],
    [J, 0, J, J],
    [J, J, 0, J],
    [J, J, J, 0]
])

# Proper Diagonalization
P, D = H.diagonalize()

print("Transformation matrix P (Eigenvectors):")
sp.pprint(P)

print("\nDiagonal matrix D (Eigenvalues):")
sp.pprint(D)
```
